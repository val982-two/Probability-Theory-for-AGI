# Парадоксы в теории вероятностей - парадокс двух конвертов

## Условие задачи
> Вам дают два абсолютно одинаковых конверта с деньгами. Известно, что в одном из них находится некоторая сумма $X$, а в другом - в два раза больше, то есть $2X$. Вы вслепую выбираете один конверт и открываете его, обнаруживая сумму $A$. Вам предлагают выбор: остаться с выбранным конвертом или поменять его на второй. Выгодно ли совершить обмен?

## В чем смысл парадокса
> На первый взгляд кажется, что во втором конверте с вероятностью $p = 0.5$ находится либо сумма $0.5A$, либо $2A$.
>
> Если вычислить математическое ожидание выигрыша при смене конверта, получим:
> $$E = 0.5 \cdot 0.5A + 0.5 \cdot 2A = 0.25A + A = 1.25A$$
>
> Так как $1.25A > A$, обмен кажется выгодным при абсолютно любой сумме $A$. Однако точно такое же рассуждение верно и для второго конверта еще до открытия первого. Возникает логическое противоречие: каждому из двух участников было бы всегда выгодно обменяться конвертами.
>
> **Математическое разрешение:** ошибка заключается в предположении, что суммы $0.5A$ и $2A$ равновероятны при фиксированном $A$. Значение $A$ - это случайная величина, зависящая от исходного $X$. В действительности теоретическое математическое ожидание суммы как в первом, так и во втором конверте одинаково и равно $1.5X$, поэтому менять конверт бессмысленно.

```python
import numpy as np

X = 100

env_1 = X
env_2 = 2 * X

p = 0.5

e_first = p * env_1 + p * env_2

e_second = p * env_2 + p * env_1

result = e_second - e_first

print(f'Теоретическое ожидание первого конверта: {e_first}')
print(f'Теоретическое ожидание второго конверта: {e_second}')
print(f'Теоретическая выгода от смены конверта: {result}')
```

```python
import random

n = 1_000_000
total_first = 0
total_second = 0

for _ in range(n):
    X = 100
    envelopes = [X, 2 * X]
    first_idx = random.choice([0, 1])
    second_idx = 1 - first_idx

    total_first += envelopes[first_idx]
    total_second += envelopes[second_idx]

mean_first = total_first / n
mean_second = total_second / n
result = mean_second - mean_first

print(f'Смоделированный средний выигрыш без смены: {mean_first}')
print(f'Смоделированный средний выигрыш со сменой: {mean_second}')
```
