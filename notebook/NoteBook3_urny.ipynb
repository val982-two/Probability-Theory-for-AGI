# Тема: Формула полной вероятности и статистическое моделирование
## Условие задачи
> В одной урне 15 белых и 5 черных шаров, а в другой — 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны — белый.

```python
import numpy as np
from scipy.special import comb

k = np.array([0, 1, 2, 3])

p_h = comb(15, k) * comb(5, 3 - k) / comb(20, 3)

p_a_given_h = (6 + k) / 10

result = np.sum(p_h * p_a_given_h)

print(f'Теоретичская вероятность: {result}')
```

```python
import random

n = 100_000
count = 0

for _ in range(n):
    urna1 = ['W'] * 15 + ['B'] * 5
    urna2 = ['W'] * 6 + ['B'] * 1

    drawn_balls = random.sample(urna1, 3)

    urna2.extend(drawn_balls)

    chosen_ball = random.choice(urna2)

    if chosen_ball == 'W':
        count += 1

result = count / n

print(f'Смоделированная вероятность: {result}')
```
