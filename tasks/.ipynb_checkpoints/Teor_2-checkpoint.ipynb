#### Задача 1.38
> Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

Пусть событие A - книги стоят слева направо в порядке нумерации томов
$$P(A) = \frac{1}{P_5} = \frac{1}{5!} = \frac{1}{120}$$

```python
import scipy.special as sc

print(round(1 / sc.factorial(5), 4))
```

#### Задача 1.43
> Наудачу взятый телефонный номер состоит из 5 цифр. Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не начинается с цифры ноль.



```python
from itertools import product, permutations

n = len([el for el in product('0123456789', repeat = 5) if el[0] != '0'])

#a - различные
m = len([el for el in permutations('0123456789', r = 5) if el[0] != '0'])
print(f'а) {round(m / n, 4)}')

#б - одинаковые
print(f'б) {round(9 / n, 4)}')

#в - нечетные
m = 0
for s in product('0123456789', repeat = 5):
    k = ''.join(s)
    if k[0] != '0':
        if all(int(el) % 2 == 1 for el in k):
            m += 1
print(f'в) {round(m / n, 4)}')
```

#### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

```python
n = len(list(product([i for i in range(1, 7)], repeat = 3)))

#a - 11 очков
m = 0
for k in product([i for i in range(1, 7)], repeat = 3):
    if sum(k) == 11: m += 1
print(f'а) {round(m / n, 4)}')

#б - выигрыш
m = 0
for k in product([i for i in range(1, 7)], repeat = 3):
    if sum(k) > 10: m += 1
print(f'б) {round(m / n, 4)}')
```
