### 1.5. Элементы комбинаторики

#### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?
> $$\text{а)} \quad \tilde{A}_{10}^5 = 10^5$$

$$\text{б)} \quad \tilde{C}_{10}^5 = C_{10+5-1}^5 = C_{14}^5$$

```python
from itertools import *
import math
import scipy.special as sc

print('itertools:')

a = len(list(product([i for i in range(1, 11)], repeat = 5)))
print(f'a) {a}')

b = len(list(combinations_with_replacement([i for i in range(1, 11)], r = 5)))
print(f'б) {b}')

print('scipy:')

a = 10**5
print(f'a) {a}')

b = sc.comb(14, 5)
print(f'б) {b}')
```

#### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?
$$P_7(3; 2; 2) = \frac{7!}{3! \, 2! \, 2!}$$

```python
k = 0
for i in product('456', repeat = 7):
    s = ''.join(i)
    if s.count('4') == 3 and s.count('5') == 2 and s.count('6') == 2:
        k += 1
print(f'itertools: {k}')

a = sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2))
print(f'scipy: {a}')
```

### 1.6. Непосредственное вычисление вероятностей

#### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки;     б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

а) Пусть событие $A$ — получение слова «ТОР»

$$P(A) = \frac{1}{A_6^3} = \frac{1}{6 \cdot 5 \cdot 4} = \frac{1}{120}$$

б) Пусть событие $B$ — получение слова «ТЕОРИЯ»

$$P(B) = \frac{1}{P_6} = \frac{1}{6!} = \frac{1}{720}$$

```python
print('itertools:')
a = len(list(permutations([i for i in 'ТЕИЯРО'], r = 3)))
print(f'a) {(1 / a):.4f}')
b = len(list(permutations([i for i in 'ТЕИЯРО'], r = 6)))
print(f'б) {(1 / b):.4f}')

print('scipy:')

a = 1 / sc.perm(6, 3)
print(f'a) {(a):.4f}')
b = 1 / sc.factorial(6)
print(f'б) {(b):.4f}')
```

#### Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Пусть событие $B$ — получение слова «АНАНАС».

$$P(B) = \frac{P_3 \cdot P_2}{P_6} = \frac{3! \, 2!}{6!} = \frac{1}{60}$$

```python
a = sc.factorial(3) * sc.factorial(2) / sc.factorial(6)
print(f'{(a):.4f}')
```

#### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

а) Пусть событие $A$ — никакие два лица одного пола не сядут рядом. 

$$P(A) = \frac{2(5!)^2}{10!}$$

б) Пусть событие $B$ — супруги, образующие пять супружеских пар, сядут рядом. 

$$P(B) = \frac{5! \, 2^5}{10!}$$

```python
a = 2 * (sc.factorial(5))**2 / sc.factorial(10)
print(f'a) {(a):.5f}')

b = sc.factorial(5) * 2**5 / sc.factorial(10)
print(f'б) {(b):.5f}')
```

#### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

а) Пусть событие $A$ — пассажиры попали в два купе. 

$$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$$

б) Пусть событие $B$ — пассажиры попали в семь купе. 

$$P(B) = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$$

в) Пусть событие $D$ — пассажиры попали в три купе.
$$P(D) = \frac{C_9^3 \left(6 C_4^4 C_4^2 C_4^1 + 3 C_4^3 C_4^3 C_4^1 + 3 C_4^3 C_4^2 C_4^2\right)}{C_{36}^7}$$

```python
a = sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7)
print(f'a) {(a):.5f}')

b = sc.comb(9, 7) * 4**7 / sc.comb(36, 7)
print(f'b) {(b):.5f}')

c = (sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1) +
    3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1) +
    3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2))
    / sc.comb(36, 7))
print(f'c) {(c):.5f}')
```
